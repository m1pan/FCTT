```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_style()
```

```python
soc_ocv = pd.read_csv('Project_3_data/SOC_OCV.csv', sep='\t')
plt.figure(figsize=(10, 6))
sns.lineplot(data=soc_ocv, x='SOC', y='Ecell/V')
plt.title('OCV vs SOC')
plt.xlabel('State of Charge (SOC)')
plt.ylabel('Open Circuit Voltage (V)')
plt.grid()
plt.show()
```

```python
testing_data = pd.read_csv('Project_3_data/Battery_Testing_Data.csv', sep=',')
plt.figure(figsize=(10, 6))
sns.lineplot(data=testing_data, x='Time (s)', y='Voltage (V)')
plt.title('Voltage vs Time')
plt.xlabel('Time (s)')
plt.ylabel('Voltage (V)')
plt.grid()
plt.show()  

plt.figure(figsize=(10, 6))
sns.lineplot(data=testing_data, x='Time (s)', y='Current (mA)')
plt.title('Current vs Time')
plt.xlabel('Time (s)')
plt.ylabel('Current (mA)')
plt.grid()
plt.show()
```
